```python
def merge_app_codes(x : pd.Series):
    return x.to_list()
```

```python
df_afspraken = df_afspraken.groupby('studyId_0831', sort=True).agg(merge_app_codes)
```

```python
def isEligible(x:pd.Series):
    if 'NFALG' in x or 'NFGEH' in x or 'NFVMI+' in x or 'NFVAL+' in x:
        return 1
    else:
        return 0
        
```

```python
df_afspraken['isEligible'] = df_afspraken.index_AfspCode.apply(isEligible)
df_afspraken
```
