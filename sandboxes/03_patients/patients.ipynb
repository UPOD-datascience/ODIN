```python
df_patients = df_patients.sort_values(by='studyId_0831')
```

```python
df_patients = df_patients.drop(['index_AfspCode', 'Agenda', 'period_start','period_end', 'Spec_Code', 'Specialisme', 'AgendaNaam', 'AgendaComment',
                                'AgendaKostenplaats', 'Description', 'AfspraakNummer', 'SubAgenda',
                                'serviceProv_Org_value', 'AfspraakStatus', 'status_code_original',
                                'status_display_original', 'type1_code', 'type1_display',
                                'type1_system', 'type2_code', 'type2_display', 'type2_system',
                                'type3_code', 'type3_display', 'type3_system','deceasedBoolean', 'deceasedDateTime', 'isTestpatient', 'index_date'],
                                axis='columns')
```

```python
df_patients.gender = df_patients.gender.apply(
    lambda g: 1 if g == 'female' else 0)
```

```python
df_patients = df_patients.sort_values(
    by='studyId_0831').groupby(by='studyId_0831').last().reset_index()
df_patients
```

---

```python
mean = df_patients.Age.mean()
std_dev = df_patients.Age.std()
median = df_patients.Age.median()
```

```python
fig = px.violin(x=['Age']*len(df_patients.Age.values), y=df_patients.Age.values, box=True,
          points='all')  # Set title, xlabel, and ylabel
fig.update_layout(
    title="Patients' Age Violin Plot",        # Title of the plot
    yaxis_title="Value",
    xaxis_title="",
    annotations=[
        dict(
            x='Age', y=mean, xref='x', yref='y',
            text=f"Mean: {mean:.2f}", showarrow=True, arrowhead=2, ax=-80, ay=-20
        ),
        dict(
            x='Age', y=median, xref='x', yref='y',
            text=f"Median: {median:.2f}", showarrow=True, arrowhead=2, ax=40, ay=-20
        ),
    ],
    width=800,  # Width in pixels
    height=600  # Height in pixels
)
```

```python
# Map binary values to categories
df_patients['Gender_Label'] = df_patients['gender'].map(
    {0: 'Male', 1: 'Female'})

# Create a pie chart
fig = px.pie(
    df_patients,
    names='Gender_Label',  # Labels for the slices
    title='Gender Distribution',  # Title of the chart
    color='Gender_Label',  # Color by gender label
    color_discrete_map={'Male': 'DarkBlue', 'Female': 'purple'}  # Custom colors
)

# Customize layout
fig.update_layout(
    title="Pie Chart of Gender Distribution",
    height=500,  # Height in pixels
    width=500    # Width in pixels
)
```
