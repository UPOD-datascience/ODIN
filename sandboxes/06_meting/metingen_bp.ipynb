Remove impossible values

```python
metingen.loc[(metingen['SystolicBP'] <=20) |
             (metingen['SystolicBP'] >= 300), 'SystolicBP']  = np.nan

metingen.loc[(metingen['DiastolicBP'] <= 20) |
             (metingen['DiastolicBP'] >= 300), 'DiastolicBP'] = np.nan
```

```python
for c in ['DiastolicBP', 'SystolicBP']:
    mu = metingen[c].mean()
    sigma = metingen[c].std()
    metingen.loc[(metingen[c] <= mu - 3*sigma) |
                 (metingen[c] >= mu + 3*sigma), c] = mu
```

```python
metingen[['DiastolicBP', 'SystolicBP']].plot(kind='box', subplots=True, figsize=(10,6))
```
