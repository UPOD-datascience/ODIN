```python
consults = pd.read_sas('consult_20240826.sas7bdat', format='sas7bdat', encoding='latin1')
consults.studyId_0831 = consults.studyId_0831.astype('int64')
consults = consults.sort_values(by=['studyId_0831', 'date'])
```

```python
def combine_texts(x):
    x_out = ""
    specialismeNaam = x['SpecialismeNaam']
    titel = x['title']
    subtitel = x['section_title_display_original']
    body = x['section_text']
    
    if pd.isna(specialismeNaam)==False:
        x_out += f"SpecialismeNaam: {specialismeNaam}\n"
    else:
        specialismeNaam=""
    
    if pd.isna(titel)==False:
        x_out += f"Onderwerp: {titel}\n"
    else:
        titel=""
    
    if pd.isna(subtitel)==False:
        if subtitel != titel:
            x_out += f"Titel: {subtitel}\n"
    else:
        subtitel=""
        
    if pd.isna(body)==False:
        x_out += f"Inhoud:{body}\n"
    else:
        return np.nan
        
    return x_out
    
```

```python
consults  = consults.assign(text=consults[['SpecialismeNaam', 
                                           'title',
                                           'section_title_display_original',
                                           'section_text']].apply(combine_texts, axis=1))
consults = consults.dropna(subset=['text'])
```

```python
consults[['studyId_0831', 'date', 'text']].to_parquet('consult_texts.parquet')
```
