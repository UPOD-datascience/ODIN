```python
import pandas as pd

import os
```

```python
root = 'T:\\lab_research\\RES-Folder-UPOD\\ODIN-UC4\\G_Output\\2_Data\\data\\whole_text_dataset\\train'

# Elenco dei file Parquet nella cartella
file_parquet = [f for f in os.listdir(root) if f.endswith('.parquet')]

# Lista per memorizzare i DataFrame letti
dataframes = []

# Leggi ogni file Parquet e aggiungilo alla lista
for file in file_parquet:
    percorso_completo = os.path.join(root, file)
    df = pd.read_parquet(percorso_completo)
    dataframes.append(df)

# Combina tutti i DataFrame in uno solo
df_combinato = pd.concat(dataframes, ignore_index=True)
```
